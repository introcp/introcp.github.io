#### Exercise Description

## Exercise 17: Alternating Split and Join

#### Exercise Description

Given the string `s = "The_quick_brown_fox"`, split the string at underscores (`_`) and store the resulting list in `split_s`.

Finally, split the reversed string by spaces and store the resulting list in `reversed_split`.

#### Your solution

```
s = "The_quick_brown_fox"
split_s = s.split('_')
print(split_s)
joined_s = ' '.join(split_s)
print(joined_s)
reversed_s = joined_s[::-1]
print(reversed_s)
reversed_split = reversed_s.split()
print(reversed_split)
# Output: ['xof', 'nworb', 'kciuq', 'ehT']
```
