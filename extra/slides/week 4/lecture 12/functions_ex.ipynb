### 4. **Simple Calculator**
Implement a function `calculator` that takes two numbers and an operator (`+`, `-`, `*`, `/`) and returns the result.

### Solution

```python
def calculator(a, b, operator):
    if operator == '+':
        return a + b
    elif operator == '-':
        return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"
    else:
        return "Invalid operator"
```

### 5. **Temperature Converter**
Write a function `celsius_to_fahrenheit` that converts Celsius to Fahrenheit.

### Solution

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
```

### 6. **Factorial Calculator (Recursive) - !!! ADVANCED !!!**
Create a recursive function `factorial` that returns the factorial of a given number. Defaults the number as 10

### Solution

```python
def factorial(n=10):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

### 7. **Check Leap Year**
Write a function `is_leap_year` that determines whether a given year is a leap year.

```python
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False
```

### 10. **Simple Interest Calculator**
Write a function `simple_interest` that calculates simple interest given principal, rate, and time.

### Solution

```python
def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100
```

### 12. **Absolute Value Function**
Create a function `absolute` that returns the absolute value of a number.

### Solution

```python
def absolute(num):
    if num >= 0:
        return num
    else:
        return -num
```

### 13. **Find the Largest Among Three Numbers**
Write a function `max_of_three` that returns the largest of three numbers.

### Solution

```python
def max_of_three(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
```

### 16. **Calculate BMI**
Write a function `calculate_bmi` that calculates Body Mass Index given weight and height.

### Solution

```python
def calculate_bmi(weight, height):
    bmi = weight / (height**2)
    return bmi
```

### 17. **Convert Minutes to Hours and Minutes**
Implement a function `convert_minutes` that takes minutes and returns hours and remaining minutes.

### Solution

```python
def convert_minutes(total_minutes):
    hours = total_minutes // 60
    minutes = total_minutes % 60
    return hours, minutes
```

### 19. **Determine Grade Based on Score**
Write a function `determine_grade` that returns a grade (`A`, `B`, `C`, `D`, `F`) based on a numerical score.

### Solution

```python
def determine_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
```

### 20. **Convert Fahrenheit to Celsius**
Implement a function `fahrenheit_to_celsius` that converts Fahrenheit to Celsius.

### Solution

```python
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9
```
