### Types of Errors in Python**

---

#### **1. Syntax Errors:**
- Occur when Python cannot understand the code due to a violation of language rules.
- Happens **before** the program runs.
- **Example:**
  ```python
  print "Hello, World"
  ```
  - **Error:** Missing parentheses in `print()`.
  - **Fix:**
    ```python
    print("Hello, World")
    ```

---

#### **2. Runtime Errors:**
- Occur when an error happens during execution, causing the program to stop.
- Happens **while** the program is running.
- **Example:**
  ```python
  x = 10
  y = 0
  z = x / y
  ```
  - **Error:** Division by zero (`ZeroDivisionError`).
  - **Fix:**
    ```python
    if y != 0:
        z = x / y
    ```

---

#### **3. Logical Errors:**
- The program runs but produces incorrect or unintended results.
- Happens **after** the program runs without errors.
- **Example:**
  ```python
  x = 10
  y = 2
  z = x - y  # Intended to add, but used subtraction
  ```
  - **Error:** The logic is wrong.
  - **Fix:**
    ```python
    z = x + y  # Corrected to addition
    ```
