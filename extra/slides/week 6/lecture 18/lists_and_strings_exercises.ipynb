### **Exercise 12: Mutating a List While Iterating**
**Description**: Explain why the following code is problematic. Suggest a fix:
```python
def remove_dups(L1, L2):
    for item in L1:
        if item in L2:
            L1.remove(item)
```

**Solution**:
The code modifies the list `L1` while iterating, which can skip elements. The fix is to iterate over a copy:

````python
def remove_dups(L1, L2):
    L1_copy = L1[:]
    for item in L1_copy:
        if item in L2:
            L1.remove(item)
    return L1
````

### **Exercise 13: Removing Duplicates**
**Description**: Write a function `remove_dups(L1, L2)` that removes all elements from `L1` that are also in `L2`.

**Solution**:

````python
def remove_dups(L1, L2):
    L1_copy = L1[:]
    for item in L1_copy:
        if item in L2:
            L1.remove(item)
    return L1


L1 = [1, 2, 3, 4]
L2 = [1, 2, 5, 6]
print(remove_dups(L1, L2))
# Output: [3, 4]
````

### **Exercise 18: List Mutation Inside a Function**
**Description**: Write a function `append_and_sort(L)` that takes a list `L`, appends a given value (e.g., 10), sorts the list, and returns it. Make sure the original list is not mutated inside the function.

**Solution**:

````python
def append_and_sort(L):
    L_copy = L[:]  # Clone the list to avoid mutation
    L_copy.append(10)
    L_copy.sort()
    return L_copy

L = [4, 2, 8, 5]
print(append_and_sort(L))  # Output: [2, 4, 5, 8, 10]
print(L)  # Output: [4, 2, 8, 5]
````

### **Exercise 19: Mutating Lists Based on Conditions**
**Description**: Write a function `remove_elements_above(L, n)` that removes all elements greater than `n` from the list `L`. Make sure to handle list mutation properly inside a loop.

**Solution**:

````python
def remove_elements_above(L, n):
    L_copy = L[:]
    for item in L_copy:
        if item > n:
            L.remove(item)
    return L

L = [1, 3, 5, 7, 9]
print(remove_elements_above(L, 5))
# Output: [1, 3, 5]
````
