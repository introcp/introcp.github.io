```python
from datetime import datetime

now = datetime.now()
print("Current date and time:", now)
```

```python
today = datetime.today()
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

```python
specific_date = datetime(2024, 12, 25, 15, 30)
print("Specific date and time:", specific_date)
```

```python
formatted_date = today.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted date:", formatted_date)
```

```python
date_string = "2024-09-17 14:30:00"
parsed_date = datetime.strptime(date_string, "%Y-%m-%d %H:%M:%S")
print("Parsed date:", parsed_date)
```

```python
from datetime import timedelta

future_date = today + timedelta(days=10)
past_date = today - timedelta(days=5)
print("Date 10 days from now:", future_date)
print("Date 5 days ago:", past_date)
```

```python
date1 = datetime(2023, 9, 1)
date2 = datetime(2024, 9, 1)
difference = date2 - date1
print("Difference between two dates:", difference.days, "days")
```

```python
current_time = now.time()
print("Current time:", current_time)
```

```python
year = 2024
is_leap = year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)
print(f"Is {year} a leap year?", is_leap)
```

```python
weekday = today.weekday()  # Monday = 0, Sunday = 6
print(f"Today is day {weekday} of the week (0=Monday, 6=Sunday)")
```

## Exercises

Write a function that takes a date of birth (as a datetime object) and returns the person's age in years. Use the current date to calculate the age.

```python
from datetime import datetime

def calculate_age(birthdate):
    today = datetime.today()
    age = today.year - birthdate.year
    # Check if the birthday has occurred this year; if not, subtract one year
    if (today.month, today.day) < (birthdate.month, birthdate.day):
        age -= 1
    return age

# Test the function
dob = datetime(1990, 4, 25)
print(f"Age: {calculate_age(dob)} years")
```

Write a function that takes a future date (as a datetime object) and calculates the number of days from today until that date.

```python
from datetime import datetime

def days_until_event(event_date):
    today = datetime.today()
    difference = event_date - today
    return difference.days

# Test the function
event = datetime(2024, 12, 25)
print(f"Days until event: {days_until_event(event)} days")
```

Write a function that takes two dates and prints all the dates between them (inclusive). Use timedelta to increment each day.

```python
from datetime import datetime, timedelta

def print_date_range(start_date, end_date):
    current_date = start_date
    while current_date <= end_date:
        print(current_date.strftime("%Y-%m-%d"))
        current_date += timedelta(days=1)

# Test the function
start = datetime(2024, 9, 15)
end = datetime(2024, 9, 20)
print_date_range(start, end)
```
