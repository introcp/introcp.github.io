### 1. **Exercise: Print numbers from 1 to N**
**Assignment:**  
Write a program that prints all numbers from 1 to N given by user using a `for` loop and `range`.

**Solution:**

```python
n = int(eval(input("Up to: ")))
for i in range(1, n + 1):
    print(i)
```

### 2. **Exercise: Sum of first N natural numbers**
**Assignment:**  
Write a program to calculate the sum of the first N (given by user) natural numbers (1 + 2 + 3 + 4 + 5) using a `while` loop.

**Solution:**

```python
i = 1
sum = 0

n = int(eval(input("Up to: ")))

while i <= n:
    sum += i
    i += 1
print("Sum:", sum)
```

### 4. **Exercise: Count down from N to 1**
**Assignment:**  
Write a program to print numbers from N (given by user) to 1 included using a `while` loop but only if N is even

**Solution:**

```python
n = int(input("From: "))

if n % 2 == 0:
    while n > 0:
        print(n)
        n -= 1
```

### 8. **Exercise: Find the factorial of 5**
**Assignment:**  
Write a program to find the factorial of 5 using a `while` loop.

**Solution:**

```python
i = 5
factorial = 1
while i > 0:
    factorial *= i
    i -= 1
print("Factorial of 5:", factorial)
```

### 11. **Exercise: Count the number of digits in a number**
**Assignment:**  
Write a program to count how many digits are in the number 123456 using a `while` loop.

**Solution:**

```python
num = 123456
count = 0
while num > 0:
    num //= 10
    count += 1
print("Number of digits:", count)
```

### 17. **Exercise: Repeat a message until a condition is met**
**Assignment:**  
Write a program that asks the user to input a string. The program should keep asking for a string until the user types "stop".

**Solution:**

```python
user_input = ""
while user_input != "stop":
    user_input = input("Enter a string (type 'stop' to end): ")
```

### 18. **Exercise: Repeat a string as many times as the length of the string**
**Assignment:**  
Write a program that asks the user to input a string and prints it as many times as the length of the string using a `for` loop.

**Solution:**

```python
user_input = input("Enter a string: ")
for i in range(len(user_input)):
    print(user_input)
```

### 19. **Exercise: Ask for a password**
**Assignment:**  
Write a program that asks the user to enter a password and keeps asking until they type the correct password ("password123").
