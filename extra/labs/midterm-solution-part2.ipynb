<p lang="IT-IT" xml:lang="IT-IT"><span lang="IT-IT" xml:lang="IT-IT" data-contrast="auto">Write a function <tt>my_function(L)</tt> that is given a list <tt>L</tt> of numbers representing coins in euro. The function should return the sum of the coins converted to dollar, where 1 euro corresponds to 1.22 dollars. </span> <span lang="IT-IT" xml:lang="IT-IT" data-contrast="auto">If there are no items in the list, the function should return 0.0 (float).</span></p>

```python
def my_function(lst):
    # special case straightaway
    if len(lst) == 0:
        return 0.0
    #
    s = 0
    for i in range(len(lst)):
        s = s + (lst[i] * 1.22)
    
    return s
```

<p>You have an integer number <tt>n</tt> as input. Define a function <tt>all_odd(n)</tt> that returns the sum of all odd integer numbers between 1 and <tt>n</tt>, <tt>n</tt> included.</p>

```python
def all_odd(n):
    s = 0 
    
    for i in range(1, n+1):
        if i % 2 != 0:
            s = s + i 
    
    return s
```

<p lang="IT-IT" xml:lang="IT-IT"><span lang="IT-IT" xml:lang="IT-IT" data-contrast="auto">Write a function <tt>my_function(L1, L2)</tt>, that takes as parameters two lists:</span></p>
<p lang="IT-IT" xml:lang="IT-IT"> </p>
<ul>
<li>each element in list <tt>L1</tt> is the name of a person;</li>
<li>each element in list <tt>L2</tt> is a character (i.e., string of length 1).</li>
</ul>
The function shall return a list <tt>L3</tt> containing persons whose name begins with a letter in <tt>L2</tt>. Elements in <tt>L3</tt> should appear in the same order in which they appear in <tt>L1</tt>.<br />
<p> </p>
</div>
<div>
<p lang="IT-IT" xml:lang="IT-IT"><span lang="IT-IT" xml:lang="IT-IT" data-contrast="auto">If <tt>L1</tt> is empty or does not </span><span lang="IT-IT" xml:lang="IT-IT" data-contrast="auto">contain persons whose name begins with a letter in <tt>L2</tt>, then <tt>L3</tt> must be an empty list.</span></p>
</div>

```python
def my_function(L1, L2):
    output = [] 
    
    for i in range(len(L1)):
        name = L1[i]
        if name[0] in L2:
            output.append(name)
    
    return output
```

```python
def my_function(lst):
    output = [ ]
    
    for i in range(len(lst)):
        if lst[i] not in output:
            output.append(lst[i])
    
    output.sort() # output = sorted(output)

    return output 
```
