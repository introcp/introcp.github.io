## Exercise 10: Traffic Light Simulator

You are tasked with creating a simple traffic light simulator in Python. The simulator should determine the next state of a traffic light based on its current color and the time that has passed.

### Requirements:

1. The program should prompt the user for two inputs:
   - The current color of the traffic light (red, yellow, or green)
   - The time passed (in seconds) since the light changed to its current color

2. The program should then determine the next color of the traffic light based on the following rules:
   - Red light duration: 60 seconds
     - If less than 60 seconds have passed, it stays red
     - After 60 seconds, it changes to green
   - Yellow light duration: 5 seconds
     - If less than 5 seconds have passed, it stays yellow
     - After 5 seconds, it changes to red
   - Green light duration: 45 seconds
     - If less than 45 seconds have passed, it stays green
     - After 45 seconds, it changes to yellow

3. The program should output the next color of the traffic light.

4. If the user inputs an invalid color, the program should handle this gracefully and output an appropriate message.

### Example Interactions:

```
Enter the current traffic light color (red, yellow, green): red
Enter the time passed in seconds: 30
The next traffic light color is: red

Enter the current traffic light color (red, yellow, green): green
Enter the time passed in seconds: 50
The next traffic light color is: yellow

Enter the current traffic light color (red, yellow, green): yellow
Enter the time passed in seconds: 10
The next traffic light color is: red

Enter the current traffic light color (red, yellow, green): blue
Enter the time passed in seconds: 5
Invalid input
```
