<p dir="ltr" style="">Write a Python function&nbsp;<tt>my_function(s,d)</tt> that, given a string <tt>s</tt> and an integer <tt>d</tt> representing the <tt>d</tt>-th day in meteorological season <tt>s</tt>, returns the month in which that day falls. The returned month should
    be a number between 1 and 12.</p><br>
<p dir="ltr" style="">You can assume that:</p>
<ol style="">
    <li dir="ltr">
        <p dir="ltr" role="presentation">The season name <tt>s</tt> is one of the strings <tt>“Autumn”</tt>, <tt>“Winter”</tt>, <tt>“Spring”</tt>, <tt>“Summer”</tt>&nbsp;</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">The input parameters are valid, i.e., that <tt>d &gt;= 1</tt> and the input season <tt>s</tt> contains at least <tt>d</tt> days</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">Spring runs from March 1 to May 31</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">Summer runs from June 1 to August 31</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">Fall (autumn) runs from September 1 to November 30</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">Winter runs from December 1 to February 28</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">You are not in a leap year (i.e., February has 28 days)</p>
    </li>
    <li dir="ltr">
        <p dir="ltr" role="presentation">Months with 30 days are April, June, September, and November</p>
    </li>
</ol>
<p></p>

```python
# Solution
def my_function(s, d):
    if s == 'Autumn':
        if d <= 30:
            return 9        # september
        elif d > 30 and d <= 30 + 31:
            return 10       # october
        else:
            return 11       # november
    elif s == 'Spring':
        if d <= 31:
            return 3        # march
        elif d > 31 and d <= 30 + 31:
            return 4        # april
        else:
            return 5        # may
    elif s == 'Summer':
        if d <= 30:
            return 6        # june
        elif d > 30 and d <= 30 + 31:
            return 7        # july
        else:
            return 8        # august
    elif s == 'Winter':
        if d <= 31:
            return 12       # december
        elif d > 31 and d <= 31 + 31:
            return 1        # january
        else:
            return 2        # february
```

<p></p><span id="docs-internal-guid-5f8c28f1-7fff-2f6e-c2e5-7bf68ee1d8d1"><p dir="ltr" style="">Write a function <tt>my_function(month,year)</tt> that takes two arguments. The first represent months and the second one the years. The <tt>month</tt> should be in
the interval [1, 12] and the <tt>year</tt> in the interval [1906, 2022]. Otherwise the function should return -1. Finally, compute the number of days in the given month. Be aware of leap years and recall the popular nursery rhyme:</p>
<p dir="ltr" style="">“30 days has September,<br>April, June, and November.<br>When short February's done<br>All the rest have 31…”</p><br>
<p dir="ltr" style="">Recall: a year is said to be a leap year, if the year is exactly divisible by 4 but not divisible by 100. Year is also a leap year if it is exactly divisible by 400.</p>
</span>

```python
# Solution
def my_function(month,year):
    # input checks
    if month < 1 or month > 12 or year < 1906 or year > 2022:
        return -1
    # is 'year' a leap year?
    leap = ((year%4 == 0)&(year%100 != 0))|(year%400 == 0)
    # get number of days
    if month == 2:
        if leap:
            return 29
        return 28
    elif (month == 4 | month == 6 | month == 9 | month== 11):
        return 30
    else:
        return 31
```

<p dir="ltr" style="">It is year 2222 and scientists have discovered how to transfer the body of a human into a robot. Write a function <tt>my_function(age, health, crime)</tt> that given an integer <tt>age</tt>, a string <tt>health</tt> representing the human's health status
    (can be “high”, “medium” or ”low”), and a boolean <tt>crime</tt> indicating whether the human has committed a crime or not returns <tt>True</tt> if the person is eligible to become a robot or <tt>False</tt> otherwise.</p>
<p dir="ltr" style="">To be eligible, a person must not have committed crimes and must not have a "low" health status. If age is between 0 and 57, health must be "high", otherwise you must have at least a "medium" state of health.</p>
<p></p>

```python
# Solution
def my_function(age, health, crime):
    if crime or health == "low":
        return False
    elif age < 58 and health == "high":
        return True
    elif age > 57 and (health == "high" or health == "medium"):
        return True
    else:
        return False
```

<p dir="ltr" style="">Write a function <tt>my_function(year, month)</tt> that takes two arguments, <tt>year</tt> and <tt>month</tt>. To be valid, the former should be in the range 1906-2022 (bounds included) and the latter must be in the usual range 1-12 (bounds included),
    otherwise the function should return -1. With valid input arguments, the function should return <tt>True</tt> if the month of a given year starts with a Monday or a Wednesday, otherwise it must return <tt>False</tt>.</p>
<p></p>

```python
# Solution
import datetime
def my_function(year, month):
    if month < 1 or month > 12 or year < 1906 or year > 2022:
        return -1
    d = datetime.date(year,month,1)
    if d.weekday() == 0 or d.weekday() == 2:
        return True
    else:
        return False
```
