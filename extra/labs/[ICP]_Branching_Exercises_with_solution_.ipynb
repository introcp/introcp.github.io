## Exercise 1: Even or Odd
Write a program that asks the user for a number and prints whether it's even or odd.

```python
# Solution
number = int(input("Enter a number: "))
if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

Create a program that converts a numerical grade to a letter grade according to the following scale:
- A: 90-100
- B: 80-89

## Exercise 3: Leap Year
Write a program that determines whether a given year is a leap year.
(A leap year is divisible by 4, but if it's divisible by 100, it must also be divisible by 400 to be a leap year.)

```python
# Solution
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

## Exercise 4: Calculator
Create a simple calculator that asks for two numbers and an operation (+, -, *, /), then performs the operation and displays the result.

```python
# Solution
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operation = input("Enter the operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero"
else:
    result = "Error: Invalid operation"

print(f"Result: {result}")
```

## Exercise 5: BMI Calculator
Write a program that calculates BMI (Body Mass Index) and categorizes it as underweight, normal weight, overweight, or obese.

```python
# Solution
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal weight"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"Your BMI is {bmi:.1f}")
print(f"Category: {category}")
```

## Exercise 6: Rock, Paper, Scissors
Implement a simple Rock, Paper, Scissors game where the user plays against the computer.

```python
# Your code here

# Solution
import random

# Generate computer's choice using random number
computer_number = random.randint(1, 3)
if computer_number == 1:
    computer_choice = "rock"
elif computer_number == 2:
    computer_choice = "paper"
else:
    computer_choice = "scissors"

user_choice = input("Enter rock, paper, or scissors: ").lower()

print(f"Computer chose: {computer_choice}")

# Check if user input is valid
if user_choice != "rock" and user_choice != "paper" and user_choice != "scissors":
    print("Invalid choice!")
elif user_choice == computer_choice:
    print("It's a tie!")
elif (user_choice == "rock" and computer_choice == "scissors") or \
     (user_choice == "paper" and computer_choice == "rock") or \
     (user_choice == "scissors" and computer_choice == "paper"):
    print("You win!")
else:
    print("Computer wins!")
```

## Exercise 7: Shipping Cost Calculator
Create a program that calculates the shipping cost for a package based on its weight and destination. The shipping company has the following rules:

Destinations are divided into three zones: Local, National, and International.
Weight is measured in kilograms and must be greater than 0.
Pricing structure:

Local:
- $1 per kg

National:

- First 5 kg: $5 per kg

- Above 5 kg: $3 per kg for additional weight

International:

- First 5 kg: $10 per kg

- Above 5 kg: $5 per kg for additional weight


For all shipments, there's a flat processing fee:

- Local: $2

- National: $5

- International: $10



Write a Python program that asks the user for the package weight and destination, then calculates and displays the total shipping cost.

```python
# Solution
# Get input from user
weight = float(input("Enter the package weight in kg: "))
destination = input("Enter the destination (Local/National/International): ").lower()

# Initialize variables
shipping_cost = 0
processing_fee = 0

# Check if weight is valid
if weight <= 0:
    print("Invalid weight. Weight must be greater than 0 kg.")
else:
    # Calculate shipping cost based on destination and weight
    if destination == "local":
        shipping_cost = weight * 1
        processing_fee = 2
    elif destination == "national":
        if weight <= 5:
            shipping_cost = weight * 5
        else:
            shipping_cost = 5 * 5 + (weight - 5) * 3
        processing_fee = 5
    elif destination == "international":
        if weight <= 5:
            shipping_cost = weight * 10
        else:
            shipping_cost = 5 * 10 + (weight - 5) * 5
        processing_fee = 10
    else:
        print("Invalid destination. Please enter Local, National, or International.")
        shipping_cost = -1  # To indicate an error

    # Calculate and display total cost if input was valid
    if shipping_cost >= 0:
        total_cost = shipping_cost + processing_fee
        print(f"Shipping cost: ${shipping_cost:.2f}")
        print(f"Processing fee: ${processing_fee:.2f}")
        print(f"Total cost: ${total_cost:.2f}")
```

## Exercise 8: Temperature Converter
Write a program that converts temperatures between Celsius and Fahrenheit based on user input.

```python
# Solution
scale = input("Enter 'C' for Celsius to Fahrenheit, or 'F' for Fahrenheit to Celsius: ").upper()
temp = float(input("Enter the temperature: "))

if scale == 'C':
    converted = (temp * 9/5) + 32
    print(f"{temp}°C is equal to {converted:.1f}°F")
elif scale == 'F':
    converted = (temp - 32) * 5/9
    print(f"{temp}°F is equal to {converted:.1f}°C")
else:
    print("Invalid input. Please enter 'C' or 'F'.")
```

## Exercise 9: Movie Ticket Price
Create a program that calculates the price of a movie ticket based on the customer's age and the day of the week.

```python
# Solution
age = int(input("Enter your age: "))
day = input("Enter the day of the week: ").lower()

base_price = 12

if age < 12:
    price = base_price * 0.5
elif age >= 65:
    price = base_price * 0.7
else:
    price = base_price

if day == "wednesday":
    price *= 0.8  # 20% discount on Wednesdays

print(f"The ticket price is ${price:.2f}")
```

## Exercise 10: Traffic Light Simulator

You are tasked with creating a simple traffic light simulator in Python. The simulator should determine the next state of a traffic light based on its current color and the time that has passed.

### Requirements:

1. The program should prompt the user for two inputs:
   - The current color of the traffic light (red, yellow, or green)
   - The time passed (in seconds) since the light changed to its current color

2. The program should then determine the next color of the traffic light based on the following rules:
   - Red light duration: 60 seconds
     - If less than 60 seconds have passed, it stays red
     - After 60 seconds, it changes to green
   - Yellow light duration: 5 seconds
     - If less than 5 seconds have passed, it stays yellow
     - After 5 seconds, it changes to red
   - Green light duration: 45 seconds
     - If less than 45 seconds have passed, it stays green
     - After 45 seconds, it changes to yellow

3. The program should output the next color of the traffic light.

4. If the user inputs an invalid color, the program should handle this gracefully and output an appropriate message.

### Example Interactions:

```
Enter the current traffic light color (red, yellow, green): red
Enter the time passed in seconds: 30
The next traffic light color is: red

Enter the current traffic light color (red, yellow, green): green
Enter the time passed in seconds: 50
The next traffic light color is: yellow

Enter the current traffic light color (red, yellow, green): yellow
Enter the time passed in seconds: 10
The next traffic light color is: red

Enter the current traffic light color (red, yellow, green): blue
Enter the time passed in seconds: 5
Invalid input
```

```python
# Solution
current_color = input("Enter the current traffic light color (red, yellow, green): ").lower()
time_passed = int(input("Enter the time passed in seconds: "))

if current_color == "red":
    if time_passed < 60:
        next_color = "red"
    else:
        next_color = "green"
elif current_color == "yellow":
    if time_passed < 5:
        next_color = "yellow"
    else:
        next_color = "red"
elif current_color == "green":
    if time_passed < 45:
        next_color = "green"
    else:
        next_color = "yellow"
else:
    next_color = "Invalid input"

print(f"The next traffic light color is: {next_color}")
```
